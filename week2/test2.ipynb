```python
import random
num = random.randint(1,10)
count = 0
flag = True
while flag:
    guess_num = int(input("请输入你猜的数字(1,10)："))
    count += 1
    if guess_num == num:
        print(f"猜对了，一共猜了{count}次。")
        flag = False
    elif guess_num > num:
        print("猜大了")
    else:
        print("猜小了")    
```

```python
# while循环嵌套案例
# 外层：坚持学习30天
# 内层，每天学习8小时
i=1
while i <= 30:
    print(f"今天是坚持学习第{i}天")

    j=1
    while j<=10:
        print(f"坚持学习第{j}小时")
        j += 1

    print("学习完成！")
    i += 1
print(f"坚持到第{i-1}天，学习结束。")
```

```python
#练习：打印九九乘法表
i = 1
while i <= 9:
    j = 1
    while j<=i:
        print(f"{j}*{i}={j*i}\t",end='')
        j += 1
    print()
    i += 1
```

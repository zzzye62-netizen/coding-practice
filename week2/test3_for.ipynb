```python
import random
account = 10000
for i in range(1,21):
    score = random.randint(1,10)
    if score >= 5:
        account -=1000 
        print(f"向员工{i}发放工资1000元，账户余额还剩余{account}元")
    else:
        print(f"员工{i}，绩效分{score}，低于5，不发工资，下一位")
    if account <=0:
        print("工资发完了，下个月领取吧")
        break
```

```python
#优化版
import random
account = 10000
for i in range(1,21):
    score = random.randint(1,10)
    if score < 5:
        print(f"员工{i}，绩效分{score}，低于5，不发工资，下一位")
        continue  # 绩效不够，跳过本次循环，下一个员工
    # 能走到这里，说明score >=5，可以发工资
    account -= 1000
    print(f"向员工{i}发放工资1000元，账户余额还剩余{account}元")
    if account <= 0:
        print("工资发完了，下个月领取吧")
        break
```

```python
#获取全部的key
my_dict ={"周杰轮":99,"林俊节":88,"张学油":77}
keys = my_dict.keys()
print(f"字典的全部keys是:{keys}")
print(type(keys))
# 遍历字典
#方式1：通过获取到全部的key来完成遍历
for name in keys:
    print(f"字典的key是:{name}")
    print(f"字典的value是:{my_dict[name]}")
#方式2：直接对字典进行for循环，每一次循环都是直接得到key
for name in my_dict:
    print(f"2字典的key是:{name}")
    print(f"2字典的value是:{my_dict[name]}")
```

```python
stiff_dict = {"王力鸿":
             {
                 "部门":"科技部",
                 "工资":3000,
                 "级别":1
             },
             "周杰轮":
             {
                 "部门":"市场部",
                 "工资":5000,
                 "级别":2
             },
            "林俊节":
            {
                "部门":"市场部",
                "工资":7000,
                "级别":3
            },
            "张学油":
            {
                "部门":"科技部",
                "工资":4000,
                "级别":1
            },
            "刘德滑":
            {
                "部门":"市场部",
                "工资":6000,
                "级别":2
            }
             }
print(f"员工的全部信息是:{stiff_dict}")
for name in stiff_dict:
    if stiff_dict[name]["级别"] == 1:
        stiff_dict[name]["级别"] += 1
        stiff_dict[name]["工资"] += 1000

print(f"全体员工级别为1的员工信息升职加薪后,信息为:{stiff_dict}")
```

```python
#案例
# 用列表 + 字典表达多个学生
students = [
    {"name": "张三", "score": 85},
    {"name": "李四", "score": 92},
    {"name": "王五", "score": 78},
]

# 1. 遍历输出每个学生的信息
for stu in students:
    print(f"{stu['name']} 的成绩是 {stu['score']}")

# 2. 用列表统计总分
scores = [stu["score"] for stu in students]
total = sum(scores)
print("总分：", total, "平均分：", total / len(scores))

# 3. 找出分数最高的学生
best = max(students, key=lambda s: s["score"])
print("最高分：", best["name"], best["score"])

# 4. 用集合去重
names = set(["张三", "李四", "张三"])
print(names)   # {'张三', '李四'}
```
